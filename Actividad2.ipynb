```python
##EJERCICIO 1

df <- read.csv("StudyArea.csv")

grafico <- df %>% 
  filter(STATE == "Idaho") %>% 
  mutate(CAUSE = ifelse(is.na(CAUSE) | trimws(CAUSE) == "", 
                        "Sin información", CAUSE)) %>% 
  select(YEAR_, CAUSE, TOTALACRES) %>% 
  group_by(YEAR_, CAUSE) %>% 
  summarise(totalacres = sum(TOTALACRES, na.rm = TRUE)) %>% 
  ggplot(mapping = aes(x = YEAR_,y = totalacres,color = CAUSE))+
  
  geom_line() +
  geom_point() +
  labs(x = 'Años', y = 'Total de acres quemados',color = 'Causa')+
  theme_bw()
```

```python
##EJERCICIO 2.1

df <- read.csv("athlete_events.csv")

df_2016 <- df %>% 
  filter(Year == 2016,
         !is.na(Medal))

tabla_medallas <- df_2016 %>% 
  count(Sport, sort = TRUE) %>% 
  head(5)

tabla_medallas %>% 
  kable(
    col.names = c("Deporte", "Número de medallas"),
    align = c("l", "c"),
    caption = "Cinco deportes con mayor número de medallas en 2016"
  ) %>% 
  kable_styling(
    bootstrap_options = c("striped", "hover", "condensed"),
    full_width = FALSE,
    position = "center"
  )
```

```python
##EJERCICIO 2.2

df_2016 <- df %>% 
  filter(Year == 2016,
         !is.na(Medal),
         Sport %in% tabla_medallas$Sport)

tabla_edad <- df_2016 %>% 
  group_by(Sport) %>% 
  summarise(
    Edad_minima = min(Age, na.rm = TRUE),
    Edad_promedio = round(mean(Age, na.rm = TRUE), 1),
    Edad_mediana = median(Age, na.rm = TRUE),
    Edad_maxima = max(Age, na.rm = TRUE)
  )

tabla_edad %>% 
  kable(
    col.names = c("Deporte", "Edad mínima", "Edad promedio", 
                  "Edad mediana", "Edad máxima"),
    align = c("l", "c", "c", "c", "c"),
    caption = "Distribución de la edad de los medallistas en 2016"
  ) %>% 
  kable_styling(
    bootstrap_options = c("striped", "hover", "condensed"),
    full_width = FALSE,
    position = "center"
  )
```

```python
##EJERCICIO 2.3

df_2016 <- df %>% 
  filter(Year == 2016,
         !is.na(Medal),
         Sport %in% tabla_medallas$Sport)

tabla_equipos <- df_2016 %>% 
  count(Team, sort = TRUE) %>% 
  head(10)

tabla_equipos %>% 
  kable(
    col.names = c("Equipo nacional", "Número de medallas"),
    align = c("l", "c"),
    caption = "Equipos nacionales con mayor número de medallas en los cinco principales deportes de 2016"
  ) %>% 
  kable_styling(
    bootstrap_options = c("striped", "hover", "condensed"),
    full_width = FALSE,
    position = "center"
  )
```

```python
##EJERCICIO 2.4

df_2016 <- df %>% 
  filter(Year == 2016,
         !is.na(Medal),
         !is.na(Weight),
         Sport %in% tabla_medallas$Sport)

tabla_peso <- df_2016 %>% 
  group_by(Sex) %>% 
  summarise(
    Peso_minimo = min(Weight, na.rm = TRUE),
    Peso_promedio = round(mean(Weight, na.rm = TRUE), 1),
    Peso_mediano = median(Weight, na.rm = TRUE),
    Peso_maximo = max(Weight, na.rm = TRUE)
  )

tabla_peso %>% 
  kable(
    col.names = c("Sexo", "Peso mínimo (kg)", "Peso promedio (kg)", 
                  "Peso mediano (kg)", "Peso máximo (kg)"),
    align = c("l", "c", "c", "c", "c"),
    caption = "Tendencia del peso de los medallistas en los cinco principales deportes de 2016"
  ) %>% 
  kable_styling(
    bootstrap_options = c("striped", "hover", "condensed"),
    full_width = FALSE,
    position = "center"
  )
```

```python
# EJERCICIO PARA ENTREGAR:
# Considere el conjunto de datos us_state_population.tsv ubicado
# en la carpeta de datos de Python de github.
# Repita el procedimiento planteado en cada item de esta sección
# para obtener el nuevo dataframe con las nuevas columnas Year y Population.
# Realice unión y separación utilizando las columnas State y Code.

df <- read_tsv("us_state_population.tsv",
               show_col_types = FALSE)

df_nuevo <- df %>% 
  pivot_longer(cols = `2010`:`2018`,
               names_to = "Year",
               values_to = "Population") %>% 
  unite("State_Code", State, Code, sep = "_") %>% 
  separate(State_Code,
           into = c("State", "Code"),
           sep = "_")

df_nuevo %>% 
  head(15) %>% 
  kable(
    col.names = c("Estado", "Código", "Año", "Población"),
    caption = "Población de los estados de Estados Unidos, 2010-2018",
    align = c("l", "c", "c", "r")
  ) %>% 
  kable_styling(
    bootstrap_options = c("striped", "hover", "condensed"),
    full_width = FALSE,
    position = "center"
  )
```

```python
# ENUNCIADO:
# Cree el histograma utilizando ggplot() con geom_histogram()
# y un tamaño de la celda de 500.
# Los datos están claramente sesgados hacia el extremo inferior
# del número de acres quemados.

datos <- read_csv("StudyArea.csv")

datos %>%
  select(ORGANIZATI, STATE, YEAR_, TOTALACRES, CAUSE) %>%
  filter(TOTALACRES >= 1000) %>%
  ggplot(mapping = aes(x = TOTALACRES)) + 
  geom_histogram(binwidth = 500,
                 fill = "steelblue",
                 color = "white") +
  coord_cartesian(xlim = c(1000, 30000)) +
  labs(
    title = "Distribución de acres quemados",
    subtitle = "Incendios forestales de 1.000 acres o más",
    x = "Total de acres quemados",
    y = "Número de incendios"
  ) +
  theme_bw()
```
