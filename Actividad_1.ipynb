## Ejercicio 1

Dado un valor `n`:

1. Calcular el factorial mediante una función.
2. Crear dos vectores aleatorios `V1` y `V2` de tamaño `n`, con valores entre 1 y 10.
3. Crear `V3`, donde `V3[i] = 0` si `V1[i] < V2[i]` y `1` en caso contrario.
4. Concatenar los tres vectores como columnas para obtener una matriz `n x 3`.


```python
# Valor de n
n = 10

# Función para calcular el factorial
def factorial(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

# Calcular factorial
resultado = factorial(n)
print(f'El factorial de {n} es: {resultado}')
```

```python
# Crear vectores aleatorios entre 1 y 10
V1 = np.random.randint(1, 11, n)
V2 = np.random.randint(1, 11, n)

# Crear V3 según la comparación
V3 = np.where(V1 < V2, 0, 1)

print('V1:', V1)
print('V2:', V2)
print('V3:', V3)
```

```python
# Convertir los vectores en columnas y concatenarlos en una matriz n x 3
matriz = np.concatenate((
    V1.reshape(-1, 1),
    V2.reshape(-1, 1),
    V3.reshape(-1, 1)
), axis=1)

df_matriz = pd.DataFrame(matriz, columns=['V1', 'V2', 'V3 (V1<V2 -> 0, si no 1)'])
df_matriz
```

**Resultado:** la matriz `n x 3` resume, fila a fila, los valores de `V1`, `V2` y la
comparación booleana codificada en `V3`. Puede verificarse visualmente que `V3` vale `0`
únicamente cuando `V1 < V2`.

## Ejercicio 2

Dados `m` (número de filas) y `n` (número de columnas):

1. Crear una matriz `A` de dimensiones `m x n` inicializada con ceros.
2. Recorrer las columnas usando `while` y llenarlas con números aleatorios enteros entre `-1` y `3`.
3. Usar dos bucles `while` para reemplazar los valores negativos por `-1` y los no negativos por `5`.
4. Repetir la transformación usando dos bucles `for` y comprobar que ambos resultados son iguales.


```python
# Valores de m y n
m = 4
n = 3

# 1. Crear matriz de ceros
A = np.zeros((m, n), dtype=int)
print('Matriz inicial:')
print(A)
```

```python
# 2. Llenar la matriz usando while
j = 0
while j < n:
    A[:, j] = np.random.randint(-1, 4, m)
    j += 1

print('Matriz con números aleatorios:')
print(A)
```

```python
# 3. Transformar usando dos bucles while
A_while = A.copy()
i = 0
while i < m:
    j = 0
    while j < n:
        if A_while[i, j] < 0:
            A_while[i, j] = -1
        else:
            A_while[i, j] = 5
        j += 1
    i += 1

print('Matriz transformada usando while:')
print(A_while)
```

```python
# 4. Repetir la transformación usando dos bucles for
A_for = A.copy()
for i in range(m):
    for j in range(n):
        if A_for[i, j] < 0:
            A_for[i, j] = -1
        else:
            A_for[i, j] = 5

print('Matriz transformada usando for:')
print(A_for)

# Comparar resultados
son_iguales = np.array_equal(A_while, A_for)
print('\n¿Las matrices finales son iguales?', son_iguales)
```

```python
# Visualización comparativa lado a lado
comparacion = pd.concat(
    [pd.DataFrame(A, columns=[f'col{j}' for j in range(n)]).add_prefix('orig_'),
     pd.DataFrame(A_while, columns=[f'col{j}' for j in range(n)]).add_prefix('while_'),
     pd.DataFrame(A_for, columns=[f'col{j}' for j in range(n)]).add_prefix('for_')],
    axis=1
)
comparacion
```

**Resultado:** la tabla comparativa muestra, para cada celda de la matriz original,
el valor obtenido con el enfoque `while` y con el enfoque `for`. Ambos métodos producen
matrices idénticas, lo que confirma que la lógica de transformación es correcta
independientemente de la estructura de control utilizada.

## Ejercicio 3

Se recolectaron datos de 8 estudiantes universitarios con las variables:
`MASA`, `ESTATURA`, `GENERO`, `ESTRATO` y `FUMA`.

Se solicita:

1. Crear un DataFrame llamado `df`.
2. Realizar un análisis descriptivo básico de todas las variables.
3. Extraer los primeros 4 estudiantes y guardarlos en `subc1`.
4. Extraer las variables categóricas y guardarlas en `subc2`.
5. Extraer las variables numéricas y guardarlas en `subc3`.


```python
# Datos de los 8 estudiantes
data = {
    'MASA': [71, 45, 50, 60, 61, 84, 55, 60],
    'ESTATURA': [183, 168, 164, 164, 167, 182, 168, 170],
    'GENERO': ['HOMBRE', 'HOMBRE', 'HOMBRE', 'MUJER',
               'HOMBRE', 'HOMBRE', 'MUJER', 'HOMBRE'],
    'ESTRATO': [2, 1, 1, 3, 3, 3, 3, 3],
    'FUMA': ['NO', 'NO', 'NO', 'SI', 'SI', 'NO', 'NO', 'SI']
}

# 1. Crear el DataFrame
df = pd.DataFrame(data)
df
```

```python
# 2. Análisis descriptivo básico de todas las variables
df.describe(include='all')
```

```python
# Información general del DataFrame
df.info()
print('\nDimensiones del DataFrame (filas, columnas):', df.shape)
```

```python
# 3. Primeros 4 estudiantes
subc1 = df.iloc[:4]
subc1
```

```python
# 4. Variables categóricas
subc2 = df[['GENERO', 'FUMA']]
subc2
```

```python
# 5. Variables numéricas
subc3 = df[['MASA', 'ESTATURA', 'ESTRATO']]
subc3
```

```python
# Visualización adicional: distribución de género y hábito de fumar
resumen_genero = df['GENERO'].value_counts()
resumen_fuma = df['FUMA'].value_counts()

print('Distribución por GÉNERO:')
print(resumen_genero)
print('\nDistribución por FUMA:')
print(resumen_fuma)
```

**Resultado:** el análisis descriptivo (`describe`) resume media, desviación estándar y
cuartiles de las variables numéricas, y frecuencias/moda de las categóricas. Los subconjuntos
`subc1`, `subc2` y `subc3` confirman que la selección por filas y por tipo de columna se
realizó correctamente.
