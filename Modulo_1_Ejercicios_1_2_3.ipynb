## Ejercicio 1

Dado un valor `n`:

1. Calcular el factorial mediante una función.
2. Crear dos vectores aleatorios `V1` y `V2` de tamaño `n`, con valores entre 1 y 10.
3. Crear `V3`, donde `V3[i] = 0` si `V1[i] < V2[i]` y `1` en caso contrario.
4. Concatenar los tres vectores como columnas para obtener una matriz `n x 3`.

```python
import numpy as np

# Valor de n
n = 10

# Función para calcular el factorial
def factorial(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

# Calcular factorial
resultado = factorial(n)
print(f'El factorial de {n} es: {resultado}')

# Crear vectores aleatorios entre 1 y 10
V1 = np.random.randint(1, 11, n)
V2 = np.random.randint(1, 11, n)

# Crear V3 según la comparación
V3 = np.where(V1 < V2, 0, 1)

# Convertir los vectores en columnas y concatenarlos
matriz = np.concatenate((
    V1.reshape(-1, 1),
    V2.reshape(-1, 1),
    V3.reshape(-1, 1)
), axis=1)

print('\nV1:', V1)
print('V2:', V2)
print('V3:', V3)
print('\nMatriz comparativa:')
print(matriz)
```

## Ejercicio 2

Dados `m` (número de filas) y `n` (número de columnas):

1. Crear una matriz `A` de dimensiones `m x n` inicializada con ceros.
2. Recorrer las columnas usando `while` y llenarlas con números aleatorios enteros entre `-1` y `3`.
3. Usar dos bucles `while` para reemplazar los valores negativos por `-1` y los no negativos por `5`.
4. Repetir la transformación usando dos bucles `for` y comprobar que ambos resultados son iguales.

```python
# Valores de m y n
m = 4
n = 3

# 1. Crear matriz de ceros
A = np.zeros((m, n), dtype=int)

print('Matriz inicial:')
print(A)

# 2. Llenar la matriz usando while
j = 0

while j < n:
    A[:, j] = np.random.randint(-1, 4, m)
    j += 1

print('\nMatriz con números aleatorios:')
print(A)

# 3. Transformar usando dos while
A_while = A.copy()
i = 0

while i < m:
    j = 0
    while j < n:
        if A_while[i, j] < 0:
            A_while[i, j] = -1
        else:
            A_while[i, j] = 5
        j += 1
    i += 1

print('\nMatriz transformada usando while:')
print(A_while)

# 4. Repetir la transformación usando dos for
A_for = A.copy()

for i in range(m):
    for j in range(n):
        if A_for[i, j] < 0:
            A_for[i, j] = -1
        else:
            A_for[i, j] = 5

print('\nMatriz transformada usando for:')
print(A_for)

# Comparar resultados
print('\n¿Las matrices finales son iguales?')
print(np.array_equal(A_while, A_for))
```

## Ejercicio 3

Se recolectaron datos de 8 estudiantes universitarios con las variables:
`MASA`, `ESTATURA`, `GENERO`, `ESTRATO` y `FUMA`.

Se solicita:

1. Crear un DataFrame llamado `df`.
2. Realizar un análisis descriptivo básico de todas las variables.
3. Extraer los primeros 4 estudiantes y guardarlos en `subc1`.
4. Extraer las variables categóricas y guardarlas en `subc2`.
5. Extraer las variables numéricas y guardarlas en `subc3`.

```python
import pandas as pd

# Datos de los 8 estudiantes
data = {
    'MASA': [71, 45, 50, 60, 61, 84, 55, 60],
    'ESTATURA': [183, 168, 164, 164, 167, 182, 168, 170],
    'GENERO': ['HOMBRE', 'HOMBRE', 'HOMBRE', 'MUJER',
               'HOMBRE', 'HOMBRE', 'MUJER', 'HOMBRE'],
    'ESTRATO': [2, 1, 1, 3, 3, 3, 3, 3],
    'FUMA': ['NO', 'NO', 'NO', 'SI', 'SI', 'NO', 'NO', 'SI']
}

# 1. Crear el DataFrame
df = pd.DataFrame(data)

print('DATAFRAME:')
display(df)
```

```python
# 2. Análisis descriptivo básico de todas las variables
print('ANÁLISIS DESCRIPTIVO:')
display(df.describe(include='all'))
```

```python
# Información general del DataFrame
print('Información del DataFrame:')
df.info()

print('\nDimensiones del DataFrame:')
print(df.shape)
```

```python
# 3. Primeros 4 estudiantes
subc1 = df.iloc[:4]

print('SUBC1 - Primeros 4 estudiantes:')
display(subc1)
```

```python
# 4. Variables categóricas
subc2 = df[['GENERO', 'FUMA']]

print('SUBC2 - Variables categóricas:')
display(subc2)
```

```python
# 5. Variables numéricas
subc3 = df[['MASA', 'ESTATURA', 'ESTRATO']]

print('SUBC3 - Variables numéricas:')
display(subc3)
```
